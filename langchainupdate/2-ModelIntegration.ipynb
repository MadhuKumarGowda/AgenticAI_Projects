### Models Integration With OPENAI, Google Gemini and GROQ


### OPENAI Model integration

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model = init_chat_model("gpt-4.1")
model
```

```python
### invoke the model

response = model.invoke("Hello, How are you ?")
### response

response.content
```

```python
### another way to use open ai

from langchain_openai import ChatOpenAI
model = ChatOpenAI(model="gpt-4.1")
response = model.invoke("Hello, How are you ?")
response
```

### Google Gemini Model Integration

```python
### Google Gemini model integration

import os

from langchain.chat_models import init_chat_model

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")

chat = init_chat_model("google_genai:gemini-3.5-flash-lite")

response = chat.invoke("How are you?")
print(response.content)
```

```python
### Another way to invoke Google Generative AI

import os
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
response = model.invoke("Why do parrots talk?")
response
```

### GROQ model integration

```python
### Using init_chat_model

import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

# Use a supported Groq model
model = init_chat_model("groq:openai/gpt-oss-120b")

response = model.invoke("Why do parrots talk?")
print(response.content)
```

```python
### Another way to use Groq

from langchain_groq import ChatGroq

model = ChatGroq(
    model="openai/gpt-oss-120b"
)

response = model.invoke("Why do parrots talk?")

print(response.content)
```
