```python
import os
from dotenv import load_dotenv
load_dotenv()

## Retrieving OPEN AI API KEY
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
### to check the version of langchain
import langchain
langchain.__version__
```

```python
from langchain.agents import create_agent

def get_weather(city:str)-> str:
    """
    Function to get city weather
    the function return some response which is called Context
    """
    return f"The weather in {city} is Sunny"

## Creating an agent with model gpt-5
agent = create_agent(
    model="gpt-5",
    tools = [get_weather], ## calling get_weather function treating as tool here
    system_prompt = "You are a helpful assistant.",   
)

agent
```

```python
### run the agent
agent.invoke({"messages":[{"role":"user", "content":"What is the weather like in Mysore?"}]})
```

```python
### to get specific output as response
response = agent.invoke({"messages":[{"role":"user", "content":"What is the weather like in Mysore?"}]})
### to get the last content which is actual message of the agent response
response["messages"][-1].content
```
