### Messages

#### Messages are the fundamentals unit of context for models in LangChain. They represents the input and output of models, carrying both the content and metadata needed to represents the state of conversation when interacting with an LLM. Messages are objects that contain:

* Role - Identifies the message type (eg. System, user)
* Content - Represents the actual content of the message (like text, images, audio, documents etc)
* Metadata - Optional fields such as response information, message IDs, and token usage

##### LangChain provides a standard message type that works across all model providers, ensuring consistent behavior  regardless of the model being called.

#### Types of Messages
  - System Message
  - AI Message :  output from model is called AI messages
  - Human Message : Human gives input to models is called Human Messages 

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:openai/gpt-oss-120b")
```

```python
# whenever we provide input to model, it treated as Human Message
model.invoke("Please tell me whats is AI") 

# Response from above model invoke operation is nothing but
# AI messages
```

#### Text Prompt

###### Text Prompts are strings: ideal for straightforward generation tasks where yiu don't need to retain conversation history.

##### Use Text Prompt when
 - We have a single standalone request
 - We don't need conversation history
 - we want minimal code complexity

#### Message Prompt

###### Alternatively, we can pass in a list of messages to the model by providing a list of message objects.

#### Message Types

 - System Message: Tells the model how to behave and provide context for interactions
 - Human Message: Represents user input and interactions with the model
 - AI Message: Responses generated by the model, including text content, tool calls and metadata
 - Tool Message: Represents the outputs of tool calls.

#### System Messages

###### A System Message represents an initial set of instructions that primes the model's behavior,  we can use a system message to set the tone, define the model's role, and establish guidelines for responses.

#### Human Messages

###### A Human messages represents user input and interactions. They can contain text, images, audio files and any other amount of multi modal content.

#### AI Messages

###### An AIMessages represents the output of a model invocation. They can include multi modal data, tool calls. and provider-specific metadata that you can later access.

#### Tool Messages

###### For models that supports tool calling, AI messages can contain tool calls. Tool messages are used to pass the results of a single tool execution back to the model


```python
# import all message types
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage("You are a code reviewer."), # Give instruction to LLM, what LLM is ?
    HumanMessage("Please review my code") # Give input to LLM to perform expected action
]

response = model.invoke(messages)
response.content
```

```python
System_message = SystemMessage("You are a helpful coding assistant.")

messages = [
    System_message,
    HumanMessage("How do i create a REST API")
]

response = model.invoke(messages)
print(response.content, end = " ", flush=True)
```

```python
## Detailed information to the LLM through system messages

sys_msg = SystemMessage("""
You are a senior Python developer with expertise in web framework.
Always provide code example and explain your reasoning.
Be concise but thorough in your explanations.
""")

messages = [
    sys_msg,
    HumanMessage("How do i create a REST API?")
]

response = model.invoke(messages)
print(response.content, end = " " , flush=True )
```

```python
### Message types examples

human_msg = HumanMessage(
    content = "Hello",
    name= "Madhu", # Optional: identify different users. # Metadata information
    id = "msg_123", # Optional: unique identifier for tracking # Metadata information
)
```

```python

response = model.invoke([
    human_msg
])
print(response.content, end= " ", flush=True)
```

```python
## Create an AI message manually
ai_msg = AIMessage("I'd be happy to help you with that questions!")

# Add a conversation History
messages = [
    SystemMessage("You are a helpful assistant"),
    HumanMessage("Can you help me?"),
    ai_msg,# Insert as if it came from the model
    HumanMessage("Great! What's 2 + 2?")
]

response = model.invoke(messages)
print(response.content)
```

```python
## To see the metadata

response.usage_metadata
```

```python
## Tool message example
from langchain.messages import ToolMessage
### After a model makes a tool call
### (Here , demonstrating manually creating the messages for brevity)

ai_msg = AIMessage(
    content=[],
    tool_calls = [{
        "name" : "get_weather",
        "args" : {"location": "San Francisco"},
        "id": "msg_123"
    }]
)

## Execute tool and create result messages
#weather_result = ,
tool_message = ToolMessage(
    content="Sunny, 72F",
    tool_call_id = "msg_123" ## Must match the call ID
)

# Continue conversation
messages = [
    HumanMessage("What's the weather in San Francisco?"),
    ai_msg,
    tool_message,
]

response = model.invoke(messages)
print(response.content)
```
